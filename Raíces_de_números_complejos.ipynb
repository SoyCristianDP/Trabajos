```python
import numpy as np
import matplotlib.pyplot as plt

def visualizar_raices(z_complejo, n):
    """
    Calcula y grafica las n raíces de un número complejo.

    """
    # 1. Calculamos módulo y argumento del número original
    r = np.abs(z_complejo)
    theta = np.angle(z_complejo)

    # 2. Calculamos el módulo de las raíces
    r_n = r ** (1/n)

    # 3. Calculamos los ángulos de cada raíz (Fórmula de De Moivre)
    angulos = [(theta + 2 * np.pi * k) / n for k in range(n)]

    # 4. Convertimos a coordenadas rectangulares (x, y) para graficar
    x_raices = [r_n * np.cos(a) for a in angulos]
    y_raices = [r_n * np.sin(a) for a in angulos]

    # Cerramos el polígono repitiendo el primer punto al final
    x_poligono = x_raices + [x_raices[0]]
    y_poligono = y_raices + [y_raices[0]]

    # --- CONFIGURACIÓN DEL GRÁFICO ---
    # En Colab, renderiza la figura inline automáticamente
    plt.figure(figsize=(8, 8))

    # Dibujamos el círculo que inscribe al polígono
    circulo = plt.Circle((0, 0), r_n, color='lightgray', fill=False, linestyle='--')
    plt.gca().add_patch(circulo)

    # Dibujamos el polígono
    plt.plot(x_poligono, y_poligono, 'b-', label=f'Polígono regular de {n} lados')

    # Graficamos los puntos de las raíces
    plt.plot(x_raices, y_raices, 'ro', markersize=8, label='Raíces')

    # Dibujamos los vectores desde el origen a cada raíz
    for x, y in zip(x_raices, y_raices):
        plt.arrow(0, 0, x, y, head_width=0.1, head_length=0.1, fc='red', ec='red',
                  length_includes_head=True, alpha=0.5)

    # Configuramos los ejes dinámicamente según el tamaño de la raíz
    limite = r_n * 1.5
    plt.xlim(-limite, limite)
    plt.ylim(-limite, limite)
    plt.axhline(0, color='black', linewidth=1)
    plt.axvline(0, color='black', linewidth=1)
    plt.grid(color='gray', linestyle='--', linewidth=0.5, alpha=0.5)

    # Etiquetas y título
    plt.title(f'Raíces {n}-ésimas de Z = {z_complejo}', fontsize=14, fontweight='bold')
    plt.xlabel('Eje Real', fontsize=12)
    plt.ylabel('Eje Imaginario', fontsize=12)
    plt.legend()

    # Mantenemos la proporción 1:1 para que los círculos no se vean ovalados
    plt.gca().set_aspect('equal', adjustable='box')

    plt.show()

# ==========================================
# PARÁMETROS MODIFICABLES
# ==========================================
# Podés cambiar estos valores y volver a darle Play a la celda

# Número complejo a analizar (recordar usar 'j' para la parte imaginaria)
numero_z = 1 + 1j

# Índice de la raíz (2 para cuadrada, 3 para cúbica, 4 para cuarta...)
indice_n = 8

# Llamamos a la función
visualizar_raices(numero_z, indice_n)
```
